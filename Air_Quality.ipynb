```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('/content/Air_Quality.csv')

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Display column names
print("Columns:", df.columns.tolist())

# --- DATA PREPARATION & AGGREGATION ---

# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'], format='mixed')

# Convert PM2.5_Level to numeric
df['PM2.5_Level'] = pd.to_numeric(df['PM2.5_Level'], errors='coerce')

# Filter data for 'city' (matching your dataset's location value)
city_center_data = df[
    df['Location'].astype(str).str.strip().str.lower() == 'city'
].copy()

# Sort data by date
city_center_data = city_center_data.sort_values('Date')

print("Filtered rows count:", len(city_center_data))

# --- CHART 1: LINE CHART ---
plt.figure(figsize=(8, 4))
plt.plot(
    city_center_data['Date'],
    city_center_data['PM2.5_Level'],
    marker='o',
    color='red'
)
plt.title('Chart 1: Air Quality Trend at City Center')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('/content/Air_Quality.csv')

# Chart 4: Histogram (Distribution of all PM2.5 readings)
plt.figure(figsize=(8, 5))
plt.hist(df['PM2.5_Level'], bins=5, color='green', edgecolor='black')
plt.title('Chart 4: PM2.5 Level Distribution')
plt.xlabel('PM2.5 Level')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# Chart 5: Horizontal Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].max().plot(kind='barh', color='teal')
plt.title('Chart 5: Maximum PM2.5 Recorded per Location')
plt.xlabel('Peak PM2.5 Level')
plt.ylabel('Location')
plt.tight_layout()
plt.show()
```

```python
# Chart 6: Vertical Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].max().plot(kind='bar', color='teal')
plt.title('Chart 6: Maximum PM2.5 Recorded per Location')
plt.xlabel('Location')
plt.ylabel('Peak PM2.5 Level')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
